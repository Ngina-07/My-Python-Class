```python
import numpy as np
```

1.**load the data**

```python
data = np.genfromtxt(
    "real_estate_dataset.csv" ,
    delimiter="," ,
    names=True,
    dtype=None ,
    encoding="utf-8"
)
```

```python
#check the shape
data.shape
```

```python
#check the field names
data.dtype.names
```

```python
#Display the first 5 records
data[:5]
```

2 **Analysing property prices**

```python
#Extracting the prices
prices = data["Price"]
```

```python
#Calculating the mean price
mean_price = np.mean(prices)
mean_price
```

```python
#calculating the median
median_price = np.median
median_price
```

```python
#calculating the minimum and maximum price
min_price = np.min(prices)
max_price = np.max(prices)
min_price
max_price
```

```python
#calculating the percentiles
percentiles = np.percentile(prices,[25,50,75])
percentiles
```

```python
#Extracting property sizes
property_size = data["Square_Feet"]
```

###Interpretation
Property prices vary across the dataset, with a difference between the minimum and maximum prices. The mean represents the average property price, while the median represents the middle price. The percentiles help show how property prices are distributed across the dataset.


3 **Analysing property sizes**

```python
#Mean property size
mean_size = np.mean(property_size)
mean_size
```

```python
#median
median_size = np.median(property_size)
median_size
```

```python
#min and max
min_size = np.min(property_size)
max_size = np.max(property_size)
max_size
min_size
```

###Interpretation
Property sizes vary across the dataset. The mean and median show the typical property size, while the minimum and maximum show the range of property sizes. This indicates that the dataset contains properties of different sizes.

4 **calculating price per square foot**

```python
#calculating for each property
price_per_sqft = prices / property_size
```

```python
#mean
mean_price_per_sqft = np.mean(price_per_sqft)
mean_price_per_sqft
```

```python
#median
median_price_per_sqft = np.median(price_per_sqft)
median_price_per_sqft
```

```python
#max and min
max_price_per_sqft = np.max(price_per_sqft)
min_price_per_sqft = np.min(price_per_sqft)
max_price_per_sqft
min_price_per_sqft
```

###Interpretation
The price per sqft shows how much buyers pay for each sqft of each property
The mean and median show the typical price per square foot, while the minimum and maximum show the range across the properties. Differences in price per square foot may reflect differences in property characteristics and location.


5 **Identify Expensive properties**

```python
#finding the 75th percentile
top_25_threshold = np.percentile(prices, 75)
top_25_threshold
```

```python
#creating the boolean mask
expensive_mask = prices >= top_25_threshold
```

```python
#counting the expensive properties
num_expensive = np.sum(expensive_mask)
num_expensive
```

```python
#Average price of expensive properties
expensive_prices = prices[expensive_mask]
average_expensive_price = np.mean(expensive_mask)
average_expensive_price
```

```python
#overall average price
overall_average_price = np.mean(prices)
average_expensive_price, overall_average_price
```

### Interpretation

The top 25% of properties represent the most expensive properties in the dataset. Their average price is higher than the overall average property price, showing that these properties are substantially more expensive than a typical property in the dataset.

6 **Applying multiple conditions**

```python
#Getting the location scores
location_score = data["Location_Score"]
```

```python
#Creating the three conditions
condition_size = property_size > mean_size
condition_price = prices >= top_25_threshold
condition_location = location_score >=8
```

```python
#combining all three
all_conditions = condition_size & condition_price & condition_location
```

```python
#counting them
num_matching = np.sum(all_conditions)
num_matching
```

```python
#display the matching properties
matching_properties = data[all_conditions]
matching_properties
```

###Interpretation
Properties were filtered using three conditions: they had to be larger than the average property size, belong to the top 25% by price, and have a location score of at least 8. Boolean masking and the & operator were used to ensure that all three conditions were satisfied. The resulting records represent properties that meet all the specified characteristics.

8 **finding the 10 most expensive properties**

```python
#sorting the price positions
sorted_indices = np.argsort(prices)
```

```python
#get the 10 most expensive
top_10_indices = sorted_indices[-10:][::-1]
```

```python
#Retrieving their information
top_10_properties = data[top_10_indices]
top_10_properties
```

```python
top_10_properties[[
    "ID",
    "Square_Feet",
    "Num_Bedrooms",
    "Location_Score",
    "Price"
]]
```

###interpretation
The property prices were sorted using np.argsort(), and the last 10 positions were selected to identify the most expensive properties. Fancy indexing was then used to retrieve their IDs, sizes, number of bedrooms, location scores, and prices. This makes it easier to examine the characteristics of the highest-priced properties.

9 **Final Analysis**

```python
#Calculate the average for the top 10
top10_size_mean = np.mean(top_10_properties["Square_Feet"])
top10_bedrooms_mean = np.mean(top_10_properties["Num_Bedrooms"])
top10_location_mean = np.mean(top_10_properties["Location_Score"])

top10_size_mean, top10_bedrooms_mean, top10_location_mean
```

```python
#calculating the overall averages
overall_size_mean = np.mean(data["Square_Feet"])
overall_bedrooms_mean = np.mean(data["Num_Bedrooms"])
overall_location_mean = np.mean(data["Location_Score"])

overall_size_mean, overall_bedrooms_mean, overall_location_mean
```

```python
#comparing the top 10 with the overall dataset
print("Top 10 average size:", top10_size_mean)
print("Overall average size:", overall_size_mean)

print("Top 10 average bedrooms:", top10_bedrooms_mean)
print("Overall average bedrooms:", overall_bedrooms_mean)

print("Top 10 average location score:", top10_location_mean)
print("Overall average location score:", overall_location_mean)
```

***Business insight***

The 10 most expensive properties are generally larger than the average property, with an average size of about 267.39 square feet compared with 174.64 square feet overall. They also have more bedrooms on average, with 4.7 bedrooms compared with approximately 2.96 bedrooms overall. Their average location score is also higher at 6.63 compared with 5.16 for the overall dataset. These results indicate that larger property size, more bedrooms, and higher location scores are characteristics associated with higher property prices in this dataset.
